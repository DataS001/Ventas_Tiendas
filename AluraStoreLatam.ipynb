```python
# Cargar los datos en DataFrames
tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)
```

```python
# Calcular el ingreso total para cada tienda
ingreso_tienda1 = tienda['Precio'].sum()
ingreso_tienda2 = tienda2['Precio'].sum()
ingreso_tienda3 = tienda3['Precio'].sum()
ingreso_tienda4 = tienda4['Precio'].sum()
```

```python
# Concatenar todos los DataFrames en uno solo
todas_las_tiendas = pd.concat([tienda, tienda2, tienda3, tienda4], ignore_index=True)
```

```python
todas_las_tiendas.to_excel('todas_las_tiendas.xlsx', index=False)
```

```python
with pd.ExcelWriter('todas_las_tiendas_separadas.xlsx', engine='openpyxl') as writer:
    tienda.to_excel(writer, sheet_name='Tienda 1', index=False)
    tienda2.to_excel(writer, sheet_name='Tienda 2', index=False)
    tienda3.to_excel(writer, sheet_name='Tienda 3', index=False)
    tienda4.to_excel(writer, sheet_name='Tienda 4', index=False)

print("Los DataFrames han sido exportados exitosamente a todas_las_tiendas_separadas.xlsx")
```

```python
import matplotlib.pyplot as plt
import matplotlib.ticker as tick
```

```python
#instalamos  NumPy en Python, que es una de las bibliotecas más populares para 
# trabajar con arreglos y operaciones matemáticas.

import numpy
```

```python
import numpy as np


tiendas = ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4']
ingresos = [ingreso_tienda1, ingreso_tienda2, ingreso_tienda3, ingreso_tienda4]

# Generar una lista de colores automáticamente
cmap = plt.get_cmap('tab20')
colores = [cmap(i / len(tiendas)) for i in range(len(tiendas))]


# Crear el gráfico de barras con colores distintos
plt.bar(tiendas, ingresos, color=colores)


# Añadir títulos y etiquetas
plt.title('Ingresos Totales por Tienda')
plt.xlabel('Tienda')
plt.ylabel('Ingreso Total (MXN)')




# Mostrar los valores encima de cada barra
#eN f'{v:,.0f} SE puede agregar la M o K al final para saber si hablamos de millones o miles de millones 
for i, v in enumerate(ingresos):
    plt.text(i, v + max(ingresos)*0.01, f'{v:,.0f} ', ha='center', va='bottom',fontsize=8)

plt.savefig('ingresos_totales_Tiendas', dpi=300)
```
