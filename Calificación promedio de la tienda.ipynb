```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

```python
print(tienda.head())
print(tienda2.head())
print(tienda3.head())
print(tienda4.head())
```

```python
#Mean para sacar el promedio 
calif_promedio_tienda1 = tienda['Calificación'].mean()
calif_promedio_tienda2 = tienda2['Calificación'].mean()
calif_promedio_tienda3 = tienda3['Calificación'].mean()
calif_promedio_tienda4 = tienda4['Calificación'].mean()
```

```python
# Imprimir promedios por tienda
print(f"Calificación promedio de la tienda 1: {calif_promedio_tienda1}")
print(f"Calificación promedio de la tienda 2: {calif_promedio_tienda2}")
print(f"Calificación promedio de la tienda 3: {calif_promedio_tienda3}")
print(f"Calificación promedio de la tienda 4: {calif_promedio_tienda4}")
```

# CALIFICACION PROMEDIO DE LAS TIENDAS

```python

import matplotlib

import pandas as pd
import matplotlib.pyplot as plt
```

```python
nombres_tiendas = ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4']
calif_promedios = [calif_promedio_tienda1, calif_promedio_tienda2, calif_promedio_tienda3, calif_promedio_tienda4]

# Crea el gráfico de barras
plt.figure(figsize=(10, 8))
bars = plt.bar(nombres_tiendas, calif_promedios, color=['blue', 'green', 'orange', 'purple'])
plt.xlabel('Tiendas')
plt.ylabel('Calificación Promedio')
plt.title('Calificación Promedio por Tienda')
plt.ylim(0, max(calif_promedios) + 1) # Ajusta los límites del eje y, si es necesario

# Añade los valores encima de las barras
for bar in bars:
    yval = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2, yval, round(yval, 2),
        ha='center', va='bottom')  # Ajusta la posición del texto

plt.show()

plt.savefig('Calificacion_promedio_tienda', dpi=300)
```
