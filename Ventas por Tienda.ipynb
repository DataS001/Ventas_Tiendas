```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

```python
# Obtener la categoría más popular (mayor cantidad) por tienda
# Agrupamos los datos por 'Producto' y sumamos los 'Precio' para cada tienda
ingresos_por_producto_tienda = tienda.groupby('Producto')['Precio'].sum()
ingresos_por_producto_tienda2 = tienda2.groupby('Producto')['Precio'].sum()
ingresos_por_producto_tienda3 = tienda3.groupby('Producto')['Precio'].sum()
ingresos_por_producto_tienda4 = tienda4.groupby('Producto')['Precio'].sum()
```

VENTAS TIENDA 1

```python
# Si aún no has convertido a DataFrame:
from matplotlib import pyplot as plt


ingresos_por_producto_tienda.plot(kind='bar', figsize=(10,6))

plt.xlabel('Producto')
plt.ylabel('Ingresos totales')
plt.title('Ingresos por producto en Tienda 1')
plt.tight_layout()
plt.show()

# Guarda la gráfica antes de mostrarla
plt.savefig("ingresos_por_producto_tienda1.png", dpi=300)  # dpi=300 para buena calidad

plt.show() #¡Nunca guardes después de mostrar, porque plt.show() vacía la figura!**

```

```python
# Supón que esta es tu Serie ya ordenada de mayor a menor
ingresos_por_producto_tienda_sorted = ingresos_por_producto_tienda.sort_values(ascending=False)

# Selecciona los 5 primeros productos más vendidos
top5 = ingresos_por_producto_tienda_sorted.head(5)

# Graficamos
top5.plot(kind='bar', figsize=(10,6), color="orange")
plt.xlabel('Producto')
plt.ylabel('Ingresos totales')
plt.title('Top 5 productos más vendidos en Tienda 1')
plt.tight_layout()

plt.savefig('Ventas_Tienda 1', dpi=300)

plt.show()
```
